## 2.1 Variables and Data Types

In Python, variables are **names bound to objects**. There is no explicit type declaration — type is determined at runtime.

### Built-in Scalar Types

| Type | Example | Description |
|------|---------|-------------|
| `int` | `42`, `-7`, `0` | Arbitrary precision integer |
| `float` | `3.14`, `-0.5` | 64-bit floating point |
| `complex` | `2+3j` | Complex number |
| `str` | `'hello'` | Immutable Unicode string |
| `bool` | `True`, `False` | Subclass of `int` (True=1, False=0) |
| `NoneType` | `None` | Singleton representing absence of value |

````python
age       = 30       # int
price     = 9.99     # float
name      = 'Alice'  # str
is_active = True     # bool
result    = None     # NoneType

# Multiple assignment
x = y = z = 0
a, b, c = 1, 2, 3          # tuple unpacking
first, *rest = [1, 2, 3, 4, 5]  # starred unpacking

print(type(age), age)
print(type(name), name)
print(f'first={first}, rest={rest}')
````

````python
# Type conversion
print(int('42'))       # 42
print(float('3.14'))   # 3.14
print(str(100))        # '100'
print(bool(0))         # False

# Truthiness rules
print(bool(''))        # False
print(bool([]))        # False
print(bool(None))      # False
print(bool('hello'))   # True
print(bool([1, 2]))    # True
````

````python
# Arbitrary precision integers
big = 10 ** 100
print(type(big))   # still int!

# Float precision caveat (IEEE 754)
print(0.1 + 0.2)   # 0.30000000000000004
from decimal import Decimal
print(Decimal('0.1') + Decimal('0.2'))  # 0.3 exact
````

## 2.2 Operators

### Arithmetic Operators

| Operator | Operation | Example | Result |
|----------|-----------|---------|--------|
| `+` | Addition | `3 + 2` | `5` |
| `-` | Subtraction | `3 - 2` | `1` |
| `*` | Multiplication | `3 * 2` | `6` |
| `/` | True division | `7 / 2` | `3.5` |
| `//` | Floor division | `7 // 2` | `3` |

````python
print(17 / 5)    # 3.4  (always float)
print(17 // 5)   # 3    (floor)
print(17 % 5)    # 2    (remainder)
print(2 ** 10)   # 1024

# Augmented assignment
n = 10
n += 5; n -= 3; n *= 2; n //= 5
print(n)  # 4
````

````python
# Comparison
x, y = 5, 10
print(x == y, x != y, x < y, x <= 5)

# Chained comparisons (unique to Python)
age = 25
print(18 <= age < 65)   # True

# Identity vs equality
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a == b)   # True  — same value
print(a is b)   # False — different objects
print(a is c)   # True  — same object
````

````python
# Logical operators + short-circuit evaluation
print(True and False)   # False
print(True or False)    # True
print(not True)         # False

# Practical: default value pattern
user_input = None
display_name = user_input or 'Anonymous'
print(display_name)
````

## 2.5 Control Flow — if / elif / else

Python uses **indentation** to define code blocks — no braces `{}`.

````python
temperature = 22

if temperature > 30:
    print('Hot')
elif temperature > 20:
    print('Warm')   # this runs
elif temperature > 10:
    print('Cool')
else:
    print('Cold')

# Ternary expression
x = 10
label = 'even' if x % 2 == 0 else 'odd'
print(label)

# match statement (Python 3.10+)
status_code = 404
match status_code:
    case 200: print('OK')
    case 404: print('Not Found')
    case 500: print('Server Error')
    case _:   print('Unknown')
````

## 2.6 Loops — for, while

Python's `for` loop iterates over any **iterable** — list, string, range, file, etc.

````python
for i in range(5):
    print(i, end=' ')   # 0 1 2 3 4
print()

# range(start, stop, step)
for i in range(2, 10, 3):
    print(i, end=' ')   # 2 5 8
print()

# enumerate — index + value
fruits = ['apple', 'banana', 'cherry']
for index, fruit in enumerate(fruits, start=1):
    print(f'{index}. {fruit}')
````

````python
# zip — iterate multiple iterables
names  = ['Alice', 'Bob', 'Charlie']
scores = [95, 87, 92]
for name, score in zip(names, scores):
    print(f'{name}: {score}')

# while loop
count = 0
while count < 5:
    print(count, end=' ')
    count += 1
print()

# break / continue / else
for n in range(10):
    if n == 3: continue
    if n == 7: break
    print(n, end=' ')
print()
````

## 2.7 Functions — Basics

Functions are **first-class objects** in Python — assignable to variables, passable as arguments, returnable from functions.

````python
def greet(name):
    """Return a greeting string."""
    return f'Hello, {name}!'

# Default arguments
def power(base, exponent=2):
    return base ** exponent

print(greet('Alice'))
print(power(3))       # 9
print(power(2, 10))   # 1024

# Keyword arguments
def describe_person(name, age, city='Unknown'):
    return f'{name}, {age} years old, from {city}'

print(describe_person(age=30, name='Bob'))
print(describe_person('Alice', 25, city='Warsaw'))
````

````python
# Multiple return values (tuple)
def min_max(numbers):
    return min(numbers), max(numbers)

low, high = min_max([3, 1, 4, 1, 5, 9, 2, 6])
print(f'min={low}, max={high}')

# Functions are objects
def apply(func, value):
    return func(value)

print(apply(str.upper, 'hello'))   # HELLO
print(apply(abs, -42))             # 42
````

## 2.8 Scope and Namespaces — LEGB Rule

Python resolves names using the **LEGB** rule:

- **L** — Local: inside the current function
- **E** — Enclosing: in any enclosing function (closures)
- **G** — Global: module-level names
- **B** — Built-in: `len`, `print`, `range`, etc.

````python
x = 'global'

def outer():
    x = 'enclosing'
    def inner():
        x = 'local'
        print(x)   # local
    inner()
    print(x)       # enclosing

outer()
print(x)           # global
````

````python
# global and nonlocal
counter = 0

def increment():
    global counter
    counter += 1

increment(); increment()
print(counter)   # 2

def make_counter():
    count = 0
    def inc():
        nonlocal count
        count += 1
        return count
    return inc

c = make_counter()
print(c(), c(), c())   # 1 2 3
````

## 3.1 Lists

A list is an **ordered, mutable sequence** of arbitrary objects.

Backed by a dynamic array: random access O(1), append O(1) amortized, insert in middle O(n).

````python
numbers = [1, 2, 3, 4, 5]
mixed   = [1, 'hello', 3.14, True, None]
nested  = [[1, 2], [3, 4], [5, 6]]

print(numbers)
print(nested[1][0])   # 3
````

````python
lst = [3, 1, 4, 1, 5, 9, 2, 6]

lst.append(7)           # add to end
lst.insert(0, 0)        # insert at index 0
lst.remove(1)           # remove first occurrence
popped    = lst.pop()   # remove last
popped_at = lst.pop(2)  # remove at index 2

print(lst)
print(f'popped={popped}, popped_at={popped_at}')
````

````python
nums = [3, 1, 4, 1, 5, 9, 2, 6]
nums.sort()               # in-place ascending
print(nums)

original    = [3, 1, 4, 1, 5]
sorted_copy = sorted(original)  # returns NEW list
print(original)     # unchanged
print(sorted_copy)

words = ['banana', 'apple', 'cherry', 'fig']
words.sort(key=len)   # sort by length
print(words)
````

````python
lst = [1, 2, 3, 4, 5]
print(len(lst), sum(lst), min(lst), max(lst))
print(lst.count(1))   # occurrences
print(lst.index(3))   # first index
print(lst[::-1])      # reversed copy

combined = lst + [6, 7, 8]
repeated = [0] * 5
print(combined)
print(repeated)
````

## 3.2 Tuples

Tuples are **ordered, immutable sequences**. Less memory than lists; usable as dict keys.

````python
point  = (3, 4)
single = (42,)          # trailing comma required for single-element!
packed = 1, 2, 3        # parentheses optional

# Immutability
try:
    point[0] = 10
except TypeError as e:
    print(e)

# Unpacking
x, y = point
first, *middle, last = (1, 2, 3, 4, 5)
print(first, middle, last)

# Swap — classic Python idiom
a, b = 10, 20
a, b = b, a
print(a, b)   # 20 10
````

````python
from collections import namedtuple

Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 4)

print(p.x, p.y)          # 3 4
print(p)                 # Point(x=3, y=4)
print(p._asdict())       # dict representation
print(isinstance(p, tuple))  # True
````

## 3.3 Dictionaries

A dictionary is a **mutable mapping of keys to values**. Since Python 3.7, dicts maintain **insertion order**.

````python
person     = {'name': 'Alice', 'age': 30, 'city': 'Warsaw'}
from_keys  = dict.fromkeys(['a', 'b', 'c'], 0)
from_pairs = dict([('x', 1), ('y', 2)])

print(person)
print(from_keys)
````

````python
d = {'name': 'Bob', 'age': 25}

# Safe read
print(d.get('city', 'N/A'))   # N/A — no KeyError

# Create / update
d['city'] = 'Krakow'
d['age']  = 26
d.setdefault('score', 100)    # adds only if missing
d.setdefault('age', 999)      # does NOT overwrite
print(d)

# Delete
del d['city']
removed = d.pop('score')
print(d, removed)
````

````python
inventory = {'apples': 50, 'bananas': 30, 'oranges': 45}

for key, value in inventory.items():
    print(f'{key}: {value}')

# Merge (Python 3.9+)
defaults  = {'color': 'red', 'size': 'M'}
overrides = {'size': 'L', 'weight': 500}
merged = defaults | overrides
print(merged)
````

## 3.4 Sets

A set is an **unordered collection of unique, hashable elements** with O(1) average membership testing.

````python
from_list  = set([1, 2, 2, 3, 3, 3])  # duplicates removed
empty_set  = set()                     # NOT {} — that's a dict!
print(from_list)

a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print(a | b)   # union
print(a & b)   # intersection
print(a - b)   # difference
print(a ^ b)   # symmetric difference
````

````python
# Deduplication
emails = ['alice@x.com', 'bob@x.com', 'alice@x.com', 'carol@x.com']
unique = list(set(emails))
print(f'Before: {len(emails)}, After: {len(unique)}')

# O(1) membership testing
allowed_ips = {'192.168.1.1', '10.0.0.1', '172.16.0.1'}
print('10.0.0.1' in allowed_ips)   # True
````

## 3.5 Comprehensions

Comprehensions are concise, readable, and generally faster than equivalent `for` + `append()` loops.

````python
# List comprehension
# Syntax: [expression for item in iterable if condition]
squares = [x**2 for x in range(10)]
evens   = [x for x in range(20) if x % 2 == 0]

matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat   = [num for row in matrix for num in row]  # nested

print(squares)
print(evens)
print(flat)
````

````python
# Dict comprehension
words        = ['apple', 'banana', 'cherry']
word_lengths = {word: len(word) for word in words}
print(word_lengths)

# Invert a dict
d        = {'a': 1, 'b': 2, 'c': 3}
inverted = {v: k for k, v in d.items()}
print(inverted)

# Set comprehension
unique_chars = {c for c in 'hello world' if c != ' '}
print(unique_chars)
````

````python
import sys

# Generator expression — lazy, uses almost no memory
gen = (x**2 for x in range(1_000_000))
print(type(gen))
print(next(gen))   # 0
print(next(gen))   # 1

total = sum(x**2 for x in range(100))  # no intermediate list
print(total)

# Memory comparison
lc = [x for x in range(10_000)]
ge = (x for x in range(10_000))
print(f'List:      {sys.getsizeof(lc):,} bytes')
print(f'Generator: {sys.getsizeof(ge):,} bytes')
````

## 3.6 `collections` Module

Specialized container types that extend built-in data structures.

````python
from collections import defaultdict, Counter, deque, namedtuple

# defaultdict — no KeyError for missing keys
word_count = defaultdict(int)
for word in 'the cat sat on the mat the cat'.split():
    word_count[word] += 1
print(dict(word_count))

# Group by first letter
by_letter = defaultdict(list)
for word in ['apple', 'banana', 'avocado', 'blueberry', 'cherry']:
    by_letter[word[0]].append(word)
print(dict(by_letter))
````

````python
# Counter — frequency counting
c = Counter('mississippi')
print(c)
print(c.most_common(3))

# Counter arithmetic
a = Counter('aabbc')
b = Counter('abcd')
print(a + b)   # combine
print(a - b)   # subtract (positive only)
print(a & b)   # intersection (min)
print(a | b)   # union (max)
````

````python
# deque — O(1) on both ends
dq = deque([1, 2, 3, 4, 5])
dq.appendleft(0)   # O(1)
dq.append(6)       # O(1)
dq.popleft()       # O(1)
dq.pop()           # O(1)
print(dq)

# Bounded deque — sliding window
recent = deque(maxlen=5)
for i in range(10):
    recent.append(i)
print(recent)   # deque([5, 6, 7, 8, 9])
````

### Best Practices — Section 3

- Use `dict.get(key, default)` for safe access.
- Prefer `Counter` over manual `defaultdict(int)` for counting.
- Use `deque` for O(1) appends/pops from both ends.

### Common Mistakes — Section 3

- `{}` creates a **dict**, not a set — use `set()` for empty sets.
- `in` on a list is O(n) — use a `set` for large membership tests.
- Modifying a list while iterating — iterate over `lst[:]` (a copy).
- `b = a[:]` is a shallow copy — nested objects are shared. Use `copy.deepcopy(a)`.

### Performance Notes — Section 3

| Append left | O(n) | O(1) | — | — |
| Index access | O(1) | O(n) | — | O(1) |
| Membership test | O(n) | O(n) | O(1) | O(1) |

### Summary — Section 3

- **List**: ordered, mutable, allows duplicates.
- **Tuple**: ordered, immutable — use for fixed data and dict keys.
- **Dict**: key-value, insertion-ordered since 3.7.

- **Comprehensions**: concise and fast — prefer over loops for transformations.
- **`collections`**: `Counter`, `defaultdict`, `deque`, `namedtuple` — always prefer over reinventing.
